```python
import pandas as pd

# Leitura do arquivo CSV
df = pd.read_csv('C:\\Users\\wagne\\Downloads\\comportamento_de_compra_real_sem_acentuacao.csv')

# Verificação de valores ausentes
print(df.isnull().sum())

# Identificação das colunas numéricas
numeric_cols = df.select_dtypes(include=['number']).columns

# Tratamento de valores ausentes (preenchendo com a média das colunas numéricas)
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# Verificação dos dados após o preenchimento
print(df.isnull().sum())
```
